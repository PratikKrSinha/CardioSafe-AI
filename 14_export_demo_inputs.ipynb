# Export test inputs for the demo
Run the cell. A zip downloads with real ECG files (.npy) and prescription images you can upload into the website.

```python
from google.colab import drive, files
import numpy as np, os, glob, shutil
drive.mount('/content/drive')
WORK = '/content/drive/MyDrive/CardioSafe'
OUT = '/content/demo_inputs'; shutil.rmtree(OUT, ignore_errors=True); os.makedirs(OUT)

# 1. ECG files: real PTB-XL test-fold recordings (never used for training), saved as .npy (1000 x 12, 100 Hz)
z = np.load(f'{WORK}/demo_examples.npz', allow_pickle=True)
X, sex, tags, truth, ids = z['X'], z['sex'], list(z['tags']), list(z['truth']), z['ecg_id']
for i in range(len(X)):
    name = f"ecg_{ids[i]}_{str(truth[i]).replace(' ', '').replace('/', '-')[:25]}_{'male' if int(sex[i]) == 0 else 'female'}.npy"
    np.save(f'{OUT}/{name}', X[i].astype('float32'))
print(len(X), 'ECG files')

# 2. Prescription images made for the demo (printed-style, synthetic)
for f in sorted(glob.glob(f'{WORK}/demo/rx_example_*.jpg')): shutil.copy(f, OUT)
print('prescription images:', len(glob.glob(f'{OUT}/*.jpg')))

shutil.make_archive('/content/demo_inputs', 'zip', OUT)
print(sorted(os.listdir(OUT))[:40])
files.download('/content/demo_inputs.zip')
```
