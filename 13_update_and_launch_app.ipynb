# CardioSafe: update the app and launch it
Run the three cells in order. Cell 2 asks you to choose the two files `cardiosafe_app.py` and `cardiosafe_view.py`.

```python
# 1. Install what the demo needs (about 1 minute)
!apt-get -qq install -y tesseract-ocr > /dev/null
!pip -q install gradio pytesseract xgboost scikit-learn scipy matplotlib requests pandas > /dev/null
print('installed')
```

```python
# 2. Mount Drive and replace the app files
from google.colab import drive, files
import os, shutil
drive.mount('/content/drive')
WORK = '/content/drive/MyDrive/CardioSafe'
assert os.path.isdir(WORK), 'MyDrive/CardioSafe not found'
up = files.upload()                       # choose cardiosafe_app.py and cardiosafe_view.py
for name in up:
    if name.startswith('cardiosafe_app') or name.startswith('cardiosafe_view'):
        clean = 'cardiosafe_app.py' if name.startswith('cardiosafe_app') else 'cardiosafe_view.py'   # handles 'cardiosafe_app (1).py'
        if os.path.exists(f'{WORK}/{clean}'): os.remove(f'{WORK}/{clean}')
        shutil.move(name, f'{WORK}/{clean}'); print('replaced', clean)
    else:
        print('ignored', name)
for need in ('cardiosafe_app.py', 'cardiosafe_view.py'):
    assert os.path.exists(f'{WORK}/{need}'), f'{need} is missing in {WORK}'
print('files ready')
```

```python
# 3. Launch the website (open the gradio.live link it prints)
import sys, importlib
sys.path.insert(0, WORK)
for m in [m for m in list(sys.modules) if m.startswith('cardiosafe_')]: del sys.modules[m]
os.environ['CARDIOSAFE_WORK'] = WORK
import cardiosafe_app as A
demo = A.build_app(WORK)
demo.launch(share=True, debug=False)
```
